```python
import json
import random


qfs_dataset = None

try:
    # Open the file in read mode ('r')
    with open('qfs_slo_test_dataset_with_generated_summaries.json', 'r', encoding='utf-8') as file:
        # Deserialize the file content into a Python dictionary
        qfs_dataset = json.load(file)

except FileNotFoundError:
    print("Error: The file 'qfs_dataset.json' was not found.")
except json.JSONDecodeError:
    print("Error: Failed to decode JSON from the file.")


print(len(qfs_dataset))

model_names_list = [
    "gpt_4_1_mini",
    "google_gemma_2_9b_it",
    "meta_llama_Meta_Llama_3_1_8B_Instruct",
    "cjvt_GaMS_9B_Instruct"
]

model_settings = [
    "decomp_aug",
    "ner_aug",
    "no_aug"
]

for model_name in model_names_list:
    for setting_name in model_settings:
        for i, q in enumerate(qfs_dataset):
            if "summaries" not in qfs_dataset[i]:
                qfs_dataset[i]["summaries"] = list()

            n_obj = {

            }
            qfs_dataset[i]["summaries"].append({
                "summary": qfs_dataset[i]["generated_summaries"][setting_name][model_name],
                "model_name": model_name,
                "setting_name": setting_name
            })

random.seed(27)


for i, q in enumerate(qfs_dataset):
    del qfs_dataset[i]["generated_summaries"]
    qfs_dataset[i]["summaries"] = random.sample(qfs_dataset[i]["summaries"], len(qfs_dataset[i]["summaries"]))


with open('qfs_slo_test_dataset_summaries_in_random_order.json', 'w', encoding='utf-8') as json_file:
    # Use json.dump() to write the data to the file
    json.dump(qfs_dataset, json_file, indent=4) 

print("Data has been stored in 'qfs_slo_test_dataset_summaries_in_random_order.json' file.")


# debuging check
# print(qfs_dataset[0])
                                               
# with open('qfs_ex3.json', 'w', encoding='utf-8') as json_file:
#     # Use json.dump() to write the data to the file
#     json.dump(qfs_dataset, json_file, indent=4) 

# print("Data has been stored in 'qfs_ex3.json' file.")
                                               
```

```python
import json
```

```python
qfs_dataset = None


try:
    # Open the file in read mode ('r')
    with open('datasets/qfs_slo_test_dataset.json', 'r', encoding='utf-8') as file:
        # Deserialize the file content into a Python dictionary
        qfs_dataset = json.load(file)

except FileNotFoundError:
    print("Error: The file 'qfs_dataset.json' was not found.")
except json.JSONDecodeError:
    print("Error: Failed to decode JSON from the file.")

for i, q in enumerate(qfs_dataset):
    qfs_dataset[i]["query"] = qfs_dataset[i]["query_candidates"][0]
    qfs_dataset[i]["id"] = i
    qfs_dataset[i]["generated_summaries"] = {
        "decomp_aug": dict(),
        "ner_aug": dict(),
        "no_aug": dict()
    }
    del qfs_dataset[i]["query_candidates"]


model_names_list = [
    "gpt_4_1_mini",
    "google_gemma_2_9b_it",
    "meta_llama_Meta_Llama_3_1_8B_Instruct",
    "cjvt_GaMS_9B_Instruct"
]

model_settings = [
    "decomp_aug",
    "ner_aug",
    "no_aug"
]

for model_name in model_names_list:
    for setting_name in model_settings:
        try:
        # Open the file in read mode ('r')
            file_name = 'generated_summaries/' + model_name + "/" + setting_name + ".json"
            with open(file_name, 'r', encoding='utf-8') as file:
                # Deserialize the file content into a Python dictionary
                s_dataset = json.load(file)
                print(file_name)
                print('Duzina s dataseta: ' + str(len(s_dataset)))
                print('Duzina qfs_dataset: ' + str(len(qfs_dataset)))
                for j, s in enumerate(s_dataset):
                    qfs_dataset[j]["generated_summaries"][setting_name][model_name] = s["summary"]
                
    
        except FileNotFoundError:
            print("Error: The file" + file_name + " was not found.")
        except json.JSONDecodeError:
            print("Error: Failed to decode JSON from the file.")

qfs_dataset
```

```python
with open('datasets/qfs_slo_test_dataset_with_generated_summaries.json', 'w', encoding='utf-8') as json_file:
    # Use json.dump() to write the data to the file
    json.dump(qfs_dataset, json_file, indent=4) 

print("Data has been stored in 'data.json' file.")
```
